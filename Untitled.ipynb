```python

# ===================================================================
# Project Title:
# Student Performance Data Analysis with Pandas, Matplotlib & Seaborn
# ===================================================================

# Objective:
# This project analyzes student performance data by performing
# data exploration, cleaning, exploratory data analysis (EDA),
# and visualization to identify patterns in marks, attendance,
# study hours, and GPA.

import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns


#load the dataset
students_data = pd.read_csv("Student_Performance_Dataset.csv")

#==================
# Data exploration
#==================

#Display the first 5 rows by default it show 5 rows 
print(students_data.head())

#Display the last 5 rows
print(students_data.tail())

#check the shape of the dataset
print(students_data.shape)

#Display the data type of each column
print(students_data.dtypes)

#Display Descriptive statistics
print(students_data.describe())

#Display info
students_data.info()

#count missing values in each columns
print(students_data.isnull().sum())

#count duplicates in each columns
print(students_data.duplicated().sum())


#================
#Data Cleaning
#================

#Reomove the duplicate from the dataset
students_data.drop_duplicates(inplace=True)

#fill the missing marks of physics by finding the mean of columns and also convert its type to int
students_data["Physics"] = students_data["Physics"].fillna(students_data["Physics"].mean()).astype(int)

#fill the missing marks of English by finding the mean of columns and also convert its type to int
students_data["English"] = students_data["English"].fillna(students_data["English"].mean()).astype(int)

#fill the missing values of attendance by finding the mean
students_data["Attendance"]=students_data["Attendance"].fillna(students_data["Attendance"].mean())

#fill the study hours by finding by mean of study hours column
students_data["StudyHours"]=students_data["StudyHours"].fillna(students_data["StudyHours"].mean())

#fill GPA missing values with mean of GPA column
students_data["GPA"]=students_data["GPA"].fillna(students_data["GPA"].mean())

#check that dataset is cleaned now
print(students_data.isnull().sum())
print(students_data.duplicated().sum())


#===================
#Data Analysis
#===================


#student which has highest GPA
highest_GPA = students_data.loc[students_data["GPA"].idxmax()]
print(highest_GPA)

#department which has highest GPA
print(students_data.groupby("Department")["GPA"].mean().nlargest(1).reset_index())

#Depratment highest Average marks
print(students_data.groupby("Department")[["Math","English","Physics"]].mean().mean(axis=1).nlargest(1).reset_index())

#average attendance of students
print(students_data[["Attendance"]].mean())

#student who studied the most hours
print(students_data.loc[students_data["StudyHours"].idxmax()])

#subject which has the highest average marks
print(students_data[["Math","English","Physics"]].mean().nlargest(1))

#Display how many students are in each department  
print(students_data.groupby("Department")["Name"].count())

#Display average GPA by Gender
print(students_data.groupby("Gender")["GPA"].mean())


#=================
#Visualization
#=================

#set the theme
sns.set_theme(style="darkgrid",palette="pastel")

#Countplot of departments
plt.figure(figsize=(8,6))
plt.title("Departments Count")
sns.countplot(data = students_data,
             x = "Department",
              hue = "Department",
              palette = "Set3"
             )
plt.show()

#Barplot of average GPA by department
plt.figure(figsize=(8,6))
plt.xlabel("Department")
plt.ylabel("Average GPA")
avg_GPA = students_data.groupby("Department")["GPA"].mean().reset_index()
plt.title("Average GPA by Department")
sns.barplot(data = avg_GPA,
           x = "Department",
            y = "GPA",
            hue="Department",
            palette="Set1"
           )
plt.show()

#Histogram of Math marks
plt.figure(figsize=(8,6))
plt.title("Math marks Distribution")
sns.histplot(data = students_data,
             x = "Math",
             bins=10,
             color="blue"
            )
plt.show()


#Boxplot of GPA by department
plt.figure(figsize=(8,6))
plt.xlabel("Department")
plt.ylabel("Average GPA")
plt.title("GPA by department")
sns.boxplot(data=students_data,
           x="Department",
            y="GPA",
            hue="Department",
            palette="Set2"
)
plt.show()

#Violinplot of GPA by department
plt.figure(figsize=(8,6))
plt.title("GPA by department")
plt.xlabel("Department")
plt.ylabel("Average GPA")
sns.violinplot(data=students_data,
              x="Department",
               y="GPA",
               hue="Department",
               palette="Set1")
plt.show()

#Scatterplot of StudyHours vs GPA
plt.figure(figsize=(8,6))
plt.title("StudyHours vs GPA")
plt.xlabel("Study Hours")
plt.ylabel("GPA")
sns.scatterplot(data=students_data,
               x="StudyHours",
                y="GPA",
                hue="Department",
                palette="Set2",
                s=30)
plt.show()

#Pairplot
sns.pairplot(data = students_data,
             hue="Department",
              palette="Set2"
)
plt.show()

#Correlation Heatmap
plt.figure(figsize=(10,8))
plt.title("Correlation Heatmap")
corr=students_data.drop(columns = "Student_ID").corr(numeric_only = True)
sns.heatmap(corr,
            annot=True,
             cmap="coolwarm",
             fmt=".2f",
             linewidth=1,
             annot_kws={"size" : 10},
             square=True,
             cbar_kws={"label" : "Correlation"}
           )
plt.xticks(rotation=30)
plt.yticks(rotation=0)
plt.savefig("Correlation_heatmap.png", dpi=300, bbox_inches="tight")
plt.show()

students_data.to_csv("Student_Performance_Cleaned.csv", index=False)

#==========================
# Key Insights / Conclusions
#==========================

# 1. AI department has the highest average GPA.
# 2. CS department has the highest number of students.
# 3. Students with more study hours generally have higher GPAs.
# 4. Average attendance is above 90%.
# 5. Mathematics has the highest average marks.
# 6. Female students have a slightly higher average GPA.
# 7. Heatmap shows a positive correlation between StudyHours and GPA.
# 8. Most students scored between 70 and 95 marks.
```
